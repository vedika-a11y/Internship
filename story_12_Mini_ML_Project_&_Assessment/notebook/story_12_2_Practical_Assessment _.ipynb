```python
# Story 12.2 – Food Delivery Delay Prediction

Build a classification model to predict whether a food delivery will be delayed or delivered on time.
```

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
```

```python
# Load Dataset

Load the Food Delivery dataset for prediction analysis.
```

```python
df = pd.read_csv("C:\\Users\\Vedika Pingat\\Downloads\\archive (10).zip")
```

```python
df.head()
```

```python
df.info()
```

```python
# Data Cleaning

Check missing values and prepare data for machine learning.
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
# Create Classification Target

Convert delivery time into delayed or on-time delivery classes.
```

```python
df['Late_Delivery'] = df['Delivery_Time_min'].apply(

    lambda x: 1 if x>45 else 0
)
```

```python
df['Late_Delivery'].value_counts()
```

```python
# Encoding Categorical Variables

Convert categorical columns into numerical format using Label Encoding.
```

```python
le = LabelEncoder()

categorical_cols = [

'Weather',
'Traffic_Level',
'Time_of_Day',
'Vehicle_Type'

]

for col in categorical_cols:

    df[col] = le.fit_transform(df[col])
```

```python
# Feature Selection

Separate input features and target variable.
```

```python
X = df.drop(

['Order_ID','Delivery_Time_min','Late_Delivery'],

axis=1

)

y = df['Late_Delivery']
```

```python
# Train Test Split

Split dataset into training and testing datasets.
```

```python
X_train,X_test,y_train,y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42
)
```

```python
# Model Selection – Random Forest

Random Forest is selected because it performs well for classification tasks and handles mixed feature types efficiently.
```

```python
model = RandomForestClassifier(

    n_estimators=100,

    random_state=42
)

model.fit(

    X_train,

    y_train
)
```

```python
# Model Prediction

Generate predictions using the trained model.
```

```python
y_pred = model.predict(X_test)
```

```python
# Accuracy Score

Evaluate overall model performance.
```

```python
accuracy = accuracy_score(

    y_test,

    y_pred
)

print("Accuracy:",accuracy)
```

```python
# Classification Report

Measure precision, recall, and F1 score.
```

```python
print(

classification_report(

    y_test,

    y_pred

))
```

```python
# Confusion Matrix

Analyse correct and incorrect predictions.
```

```python
print(

confusion_matrix(

    y_test,

    y_pred

))
```

```python
# Feature Importance

Identify features influencing delivery delay prediction.
```

```python
importance = pd.DataFrame({

'Feature':X.columns,

'Importance':model.feature_importances_

})

print(

importance.sort_values(

    by='Importance',

    ascending=False
))
```
