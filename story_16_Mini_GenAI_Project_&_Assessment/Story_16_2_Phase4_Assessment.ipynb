```python
# Import required libraries
```

```python
import re
import string
from textblob import TextBlob
import google.generativeai as genai
```

```python
# Sample paragraph for analysis
```

```python
text = """
Artificial Intelligence (AI) is transforming industries worldwide by automating repetitive tasks, improving decision-making, and increasing productivity. Businesses use AI in healthcare, banking, education, manufacturing, and customer service to analyze data and improve efficiency. AI-powered systems can detect fraud, recommend products, assist doctors in diagnosis, and automate routine work. While AI creates new opportunities and innovations, it also raises concerns about job displacement, data privacy, and ethical decision-making. Governments and organizations are working together to ensure responsible AI development. As technology advances, AI is expected to become an essential part of everyday life, helping people solve complex problems more efficiently.
"""
```

```python
# Display original text
```

```python
print("Original Text:\n")
print(text)
```

```python
# Clean the text
```

```python
def clean_text(text):

    text = text.lower()

    text = re.sub(r"http\S+", "", text)

    text = re.sub(r"\d+", "", text)

    text = text.translate(str.maketrans('', '', string.punctuation))

    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
cleaned_text = clean_text(text)

print(cleaned_text)
```

```python
# Perform sentiment analysis
```

```python
blob = TextBlob(cleaned_text)

polarity = blob.sentiment.polarity

print("Polarity:", polarity)
```

```python
if polarity > 0:
    sentiment = "Positive"
elif polarity < 0:
    sentiment = "Negative"
else:
    sentiment = "Neutral"

print("Sentiment:", sentiment)
```

```python
# Connect to Gemini API
```

```python
genai.configure(api_key="YOUR_API_KEY")
```

```python
model = genai.GenerativeModel("gemini-2.5-flash")
```

```python
# Generate AI summary
```

```python
prompt = f"""
Summarize the following text in 3 to 4 sentences:

{text}
"""

try:
    response = model.generate_content(prompt)
    summary = response.text
    print(summary)

except Exception as e:
    print("Error occurred:")
    print(e)

```

```python
print("="*70)
print("FINAL NLP ANALYSIS")
print("="*70)

print("\nOriginal Text")
print(text)

print("\nCleaned Text")
print(cleaned_text)

print(f"\nPolarity Score : {polarity:.2f}")

print(f"\nSentiment : {sentiment}")

print("\nAI Generated Summary")
print(summary)

print("\nAnalysis Completed Successfully")
```
