```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
df = pd.read_csv(r"C:\Users\Vedika Pingat\Intern-Projects\story_11_model_improvement\data\online_shoppers_intention.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
le = LabelEncoder()

categorical_columns = [
    'Month',
    'VisitorType',
    'Weekend',
    'Revenue'
]

for col in categorical_columns:
    df[col] = le.fit_transform(df[col])
```

```python
X = df.drop("Revenue", axis=1)

y = df["Revenue"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=100
)
```

```python
log_model = LogisticRegression(max_iter=1000)

log_model.fit(X_train, y_train)

y_pred_log = log_model.predict(X_test)
```

```python
log_accuracy = accuracy_score(y_test, y_pred_log)

log_precision = precision_score(y_test, y_pred_log)

log_recall = recall_score(y_test, y_pred_log)

log_f1 = f1_score(y_test, y_pred_log)

print("Accuracy:", log_accuracy)
print("Precision:", log_precision)
print("Recall:", log_recall)
print("F1 Score:", log_f1)
```

```python
rf_model = RandomForestClassifier(random_state=100)

rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)
```

```python
rf_accuracy = accuracy_score(y_test, y_pred_rf)

rf_precision = precision_score(y_test, y_pred_rf)

rf_recall = recall_score(y_test, y_pred_rf)

rf_f1 = f1_score(y_test, y_pred_rf)

print("Accuracy:", rf_accuracy)
print("Precision:", rf_precision)
print("Recall:", rf_recall)
print("F1 Score:", rf_f1)
```

```python
comparison_df = pd.DataFrame({

    "Model": [
        "Logistic Regression",
        "Random Forest"
    ],

    "Accuracy": [
        log_accuracy,
        rf_accuracy
    ],

    "Precision": [
        log_precision,
        rf_precision
    ],

    "Recall": [
        log_recall,
        rf_recall
    ],

    "F1 Score": [
        log_f1,
        rf_f1
    ]
})

comparison_df
```

```python
## Model Comparison Summary

Two machine learning models were trained and compared using the Online Shoppers Purchasing Intention dataset.

Random Forest performed better than Logistic Regression because it achieved higher accuracy and a stronger F1-score. It was better at identifying customer purchase behavior patterns.

Therefore, Random Forest was selected as the better model for this task.
```
