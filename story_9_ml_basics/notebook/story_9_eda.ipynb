```python
# Story 9.2 – Problem Understanding and Data Cleaning
```

```python
target = "Survived"
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df = df.drop(columns=["Cabin"], errors="ignore")
```

```python
selected_columns = ["Survived", "Pclass", "Sex", "Age", "SibSp", "Parch", "Fare", "Embarked"]
df = df[selected_columns]
```

```python
## Encoding Categorical Variables
```

```python
df_encoded = pd.get_dummies(df, columns=["Sex", "Embarked"], drop_first=True)
df_encoded.head()
```

```python
df_encoded.info()
```

```python
df_encoded["Sex_male"] = df_encoded["Sex_male"].astype(int)
df_encoded["Embarked_Q"] = df_encoded["Embarked_Q"].astype(int)
df_encoded["Embarked_S"] = df_encoded["Embarked_S"].astype(int)
```
