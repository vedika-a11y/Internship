```python
# Story 14.1 – Text Summarization using Google Gemini API

## Objective

The objective of this project is to explore prompt engineering techniques using the Google Gemini API. Different prompt styles are tested 
to observe how prompt design influences the quality, structure, and tone of AI-generated summaries.

```

```python
# Import the required library
```

```python
import google.generativeai as genai
```

```python
# Configure the Gemini API
```

```python
API_KEY = "YOUR_API_KEY"

genai.configure(api_key=API_KEY)
```

```python
# Load the Gemini model
```

```python
model = genai.GenerativeModel("models/gemini-2.5-flash")
```

```python
response = model.generate_content("Hello")

print(response.text)
```

```python
## Prompt Experiment 1 – General Prompt
#The purpose of this experiment is to observe how the AI responds to a broad and open-ended prompt.
#A general prompt provides minimal instructions, allowing the model to generate a comprehensive response based on its understanding of the topic.
```

```python
prompt_1 = """
Explain how Artificial Intelligence is transforming software development.
"""

response = model.generate_content(prompt_1)

print(response.text)
```

```python
### Observation
#The general prompt produced a detailed and informative response. Since no specific format or audience was mentioned, the model generated a broad explanation covering multiple aspects of software development.
```

```python
## Prompt Experiment 2 – Structured Response
#In this experiment, the prompt includes formatting instructions.
#The objective is to observe how the AI modifies its response when a specific structure is requested.
```

```python
prompt_2 = """
Explain how Artificial Intelligence is transforming software development.

Provide the answer in exactly five bullet points.
"""

response = model.generate_content(prompt_2)

print(response.text)
```

```python
#observation
#The AI generated a concise response in the requested bullet-point format.
#Compared to the previous prompt, the response was shorter, more structured, and focused on the most important applications of Artificial Intelligence in software development.
#The model successfully followed the formatting instruction and presented the information in an easy-to-read format.
```

```python
## Prompt Experiment 3 – Role-Based Prompt
#In this experiment, the AI is assigned a professional role and a target audience.
```

```python
prompt_3 = """
You are a senior technology consultant.

Explain how Artificial Intelligence is transforming software development for a Chief Executive Officer (CEO).

Limit the response to approximately 100 words.
"""

response = model.generate_content(prompt_3)

print(response.text)
```

```python
## Output Analysis

#The AI generated a concise business-oriented explanation targeted at a Chief Executive Officer (CEO).
#Unlike the previous responses, the output focused on organizational benefits such as productivity improvement, faster software delivery, reduced development costs, innovation, and competitive advantage rather than technical implementation details.
#The model successfully adapted its language and content to match the specified audience.
```
