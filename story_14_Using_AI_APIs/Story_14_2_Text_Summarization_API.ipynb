```python
# Select Customer Reviews for Analysis
#For this experiment, the first 10 reviews are selected. These reviews will be summarized using the Gemini API and compared with the original text.
```

```python
reviews = df['Review.Text'].dropna().head(10)

reviews
```

```python
for i, review in enumerate(reviews, start=1):
    print(f"Review {i}:")
    print(review)
    print("-" * 80)
```

```python
## Generate AI Summaries
#The Gemini API is used to generate concise summaries of customer reviews.
#Each review is sent to the AI model, which produces a short summary highlighting the key customer opinion.
#This helps businesses quickly understand customer feedback without reading every review in full.
```

```python
print(len(reviews))
```

```python
all_reviews = "\n\n".join(
    [f"Review {i+1}: {review}" for i, review in enumerate(reviews)]
)

prompt = f"""
You are a customer feedback analyst.

Summarize each of the following 10 customer reviews in one short sentence.

{all_reviews}
"""

response = model.generate_content(prompt)

print(response.text)
```

```python
batch_summary = response.text

print(batch_summary)
```
