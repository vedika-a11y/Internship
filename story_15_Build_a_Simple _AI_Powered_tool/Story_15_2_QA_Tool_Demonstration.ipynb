```python
## Business Use Case
Insurance companies receive many customer queries regarding policy coverage, claim procedures, exclusions, waiting periods, and policy renewals.
Instead of reading the entire insurance document, customers can ask questions directly to an AI-powered assistant.
This project demonstrates a simple AI-powered insurance policy assistant.
```

```python
## Import Required Libraries
 #Gemini API to generate answers.
import google.generativeai as genai
```

```python
## Configure Gemini API
#The Gemini API key is used to authenticate requests and access the AI model.
genai.configure(api_key="YOUR_API_KEY")

model = genai.GenerativeModel("gemini-2.5-flash")

print("Gemini Model Loaded Successfully")
```

```python
#load document 
with open(
    r"C:\Users\Vedika Pingat\Intern-Projects\insurance_policy.txt",
    "r",
    encoding="utf-8"
) as file:
    document = file.read()

print("Document Loaded Successfully")
```

```python
print(document[:500])
```

```python
## Demonstration of the Q&A Tool
#The following examples demonstrate how the AI-powered Q&A system answers questions using information from the insurance policy document.


questions = [
    "What is the coverage amount?",
    "What documents are required for a claim?",
    "Does the policy cover cosmetic surgery?",
    "What is the waiting period?",
    "How can I renew my policy?"
]

questions
```

```python
#answers from AI

for question in questions:

    prompt = f"""
    You are an insurance policy assistant.

    Insurance Policy:
    {document}

    Question:
    {question}

    Answer only using the information provided in the policy document.
    """

    response = model.generate_content(prompt)

    print("=" * 70)
    print("QUESTION:")
    print(question)

    print("\nANSWER:")
    print(response.text)

    print("\n")
```

```python
## Strengths of the Tool

1. Easy to build and understand.
2. Provides quick answers from the insurance policy document.
3. Reduces the need to manually search through long documents.
4. Uses AI to understand natural language questions.
5. Can be adapted for insurance, banking, HR policies, healthcare, and customer support applications.
6. Demonstrates practical usage of Large Language Models (LLMs).
```

```python
## Limitations of the Tool

1. The tool can only answer questions based on the provided document.
2. It cannot verify information outside the document.
3. The quality of answers depends on the quality of the document.
4. Large documents may become difficult to process efficiently.
5. This is a simple prototype and does not use advanced retrieval techniques.
```
