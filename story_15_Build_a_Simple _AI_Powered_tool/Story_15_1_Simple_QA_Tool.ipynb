```python
import google.generativeai as genai

genai.configure(api_key="YOUR_API_KEY")

model = genai.GenerativeModel("gemini-2.5-flash")

# Read policy document
with open(
    r"C:\Users\Vedika Pingat\Intern-Projects\insurance_policy.txt",
    "r",
    encoding="utf-8"
) as file:
    document = file.read()

# User question
question = input("Ask a question about the insurance policy: ")

# Prompt
prompt = f"""
You are an insurance policy assistant.

Insurance Policy:
{document}

Question:
{question}

Answer only using the information provided in the policy document.
"""

# Generate response
response = model.generate_content(prompt)

print("\nAnswer:")
print(response.text)
```
